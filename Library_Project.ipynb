**Import Statements**

```python
import csv
import os

filename = "library.csv"
fields = ['ID', 'Title', 'Author', 'Genre', 'Year', 'Location']

with open(filename, 'w', newline='') as csvfile:
    writer = csv.DictWriter(csvfile, fieldnames=fields)
    writer.writeheader()

data = []
new_additions = []
current_id = 1

print(f"File '{filename}' created and system initialized.")
```

**Function to add a record**

```python
def addRecord():
    global current_id
    title = input("Enter Book Title: ")
    author = input("Enter Author: ")
    genre = input("Enter Genre: ")
    year = input("Enter Year: ")
    location = input("Enter Location: ")

    record = {
        'ID': current_id,
        'Title': title,
        'Author': author,
        'Genre': genre,
        'Year': year,
        'Location': location
    }

    new_additions.append(record)
    data.append(record)
    current_id += 1
    print("-" * 15)
    print("New Record added successfully!")
```

**Function to display the data**

```python
def displayData():
    if not data:
        print("\nNo records found in the system.")
        return

    print("\n" + "="*95)
    for item in fields:
        print(f"%-15s" % item, end='')
    print("\n" + "="*95)

    for row in data:
        for key, val in row.items():
            print(f"%-15s" % val, end='')
        print()
    print("="*95)
```

**Function to Search and display data**

```python
def searchData():
  #Search the data
    search_term = input("What would you like to search for? ").lower()
    results = []

    for row in data:
        for key, val in row.items():
            if search_term in str(val).lower():
                results.append(row)
                break

   #Display the results
    if results:
        print("\nSearch Results:")
        print("="*95)
        for item in fields:
            print(f"%-15s" % item, end='')
        print("\n" + "="*95)
        for item in results:
            for key, val in item.items():
                print(f"%-15s" % val, end='')
            print()
        print("="*95)
    else:
        print(f"\nSorry, no records found for '{search_term}'")
```

**Execute the Code (The Menu)**

```python
print("------Welcome to the Library Management System------")

while True:
    print("\n--- What would you like to do? ---")
    print("1 - Add a book")
    print("2 - Display your books")
    print("3 - Search for a Book")
    print("x - Exit and Save")

    choice = input("> ").lower()

    if choice == '1':
        addRecord()
    elif choice == '2':
        displayData()
    elif choice == '3':
        searchData()
    elif choice == 'x':
        print("Thank you! Shutting down and preparing to save...")
        break
    else:
        print("Sorry, I didn't recognize that option.")
```

**Save the newly added record in the CSV**

```python
if len(new_additions) > 0:
   # Adding data with append mode ('a')
    with open(filename, 'a', newline='') as csvfile:
        writer = csv.DictWriter(csvfile, fieldnames=fields)
        for item in new_additions:
            writer.writerow(item)

    print(f"\nSUCCESS: {len(new_additions)} records saved to {filename}!")
    new_additions = []
else:
    print("\nNothing new to save.")
```

**list the number of authers per genre**

```python
genre_counts = {}
for book in data:
    g = book['Genre']
    genre_counts[g] = genre_counts.get(g, 0) + 1

print("\n--- Statistics: Authors per Genre ---")
for genre, count in genre_counts.items():
    print(f"{genre}: {count}")
```
